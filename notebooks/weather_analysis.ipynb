# 🌦️ Weather Data Analysis Notebook

Welcome to the **Weather Data Analysis** notebook! In this notebook, we analyze daily weather patterns including temperature, humidity, rainfall, wind speed, and weather conditions using **Pandas** and **Matplotlib**.

## Step 1: Import Libraries and Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the weather dataset
df = pd.read_csv("../data/weather.csv")

print("Dataset loaded successfully! First 5 rows:")
df.head()
```

## Step 2: Statistical Summary Table

```python
# Display summary statistics for numerical columns
df.describe()
```

## Step 3: Weather Condition Grouping
Group weather data by condition (`Sunny`, `Rainy`, `Cloudy`) to inspect mean temperature, humidity, rainfall, and count of days.

```python
# Weather condition grouping summary
weather_summary = df.groupby("weather").agg(
    days_count=("date", "count"),
    avg_temperature=("temperature", "mean"),
    avg_humidity=("humidity", "mean"),
    total_rainfall=("rainfall", "sum"),
    avg_wind_speed=("wind_speed", "mean")
).round(2)

weather_summary
```

## Step 4: Temperature & Humidity Trend Analysis

```python
plt.figure(figsize=(12, 5))
plt.plot(df["date"], df["temperature"], marker="o", color="orange", label="Temperature (°C)", linewidth=2)
plt.plot(df["date"], df["humidity"], marker="s", color="teal", label="Humidity (%)", linewidth=2, linestyle="--")

plt.title("Daily Temperature & Humidity Trends", fontsize=14, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

## Step 5: Daily Rainfall Bar Chart

```python
plt.figure(figsize=(12, 5))
plt.bar(df["date"], df["rainfall"], color="royalblue", edgecolor="navy", alpha=0.8)

plt.title("Daily Rainfall (mm)", fontsize=14, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Rainfall (mm)")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

## Step 6: Average Metrics by Weather Condition

```python
weather_summary[["avg_temperature", "avg_humidity"]].plot(
    kind="bar", figsize=(8, 5), color=["coral", "cadetblue"]
)
plt.title("Average Temperature & Humidity by Weather Condition", fontsize=12, fontweight="bold")
plt.ylabel("Value")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
